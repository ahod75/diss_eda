```python
import gurobipy as gp
from gurobipy import GRB

import sys
from pyprojroot import here

ROOT_DIR = here()
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))
```

```python
# 1. Initialize the model
model = gp.Model("continuous_lp")

# 2. Add variables (Default is continuous, lower bound 0)
x = model.addVar(ub=10.0, name="x")
y = model.addVar(ub=10.0, name="y")

# 3. Set objective
model.setObjective(3 * x + 4 * y, GRB.MAXIMIZE)

# 4. Add constraints
model.addConstr(x + 2 * y <= 14, "c0")
model.addConstr(3 * x - y >= 0, "c1")

# 5. Solve the model
model.optimize()

# 6. Extract the results
if model.Status == GRB.OPTIMAL:
    print("\n--- Optimal Solution Found ---")
    for v in model.getVars():
        print(f"{v.VarName}: {v.X}")
    print(f"Objective Value: {model.ObjVal}")
```

```python
# 1. Initialize the model
model = gp.Model("knapsack_mip")

# Problem Data
weights = [2, 3, 4, 5, 9]
values = [3, 4, 8, 8, 10]
capacity = 14
num_items = len(weights)

# 2. Add binary variables (0 = leave item, 1 = pack item)
# model.addVars creates a dictionary of variables, which is highly efficient
x = model.addVars(num_items, vtype=GRB.BINARY, name="pack")

# 3. Set objective (Maximize total value of packed items)
# gp.quicksum is Gurobi's highly optimized version of Python's sum()
model.setObjective(
    gp.quicksum(values[i] * x[i] for i in range(num_items)), 
    GRB.MAXIMIZE
)

# 4. Add constraint (Total weight must be <= capacity)
model.addConstr(
    gp.quicksum(weights[i] * x[i] for i in range(num_items)) <= capacity, 
    "weight_limit"
)

# 5. Solve the model
model.optimize()

# 6. Extract the results
if model.Status == GRB.OPTIMAL:
    print("\n--- Optimal Packing Plan ---")
    for i in range(num_items):
        if x[i].X > 0.5: # Checking if binary variable was set to 1
            print(f"Pack item {i} (Weight: {weights[i]}, Value: {values[i]})")
    print(f"Total Value Packed: {model.ObjVal}")
```
